```python
import os
```

```python
import json
with open('/content/Pages.json', 'r') as f:
    data = json.load(f)

annotations = data['_via_img_metadata']
```

```python
import cv2
import numpy as np

def create_mask(image_shape, regions):
    mask = np.zeros(image_shape[:2], dtype=np.uint8)
    for region in regions:
        shape = region['shape_attributes']
        if shape['name'] == 'rect':
            x, y, width, height = shape['x'], shape['y'], shape['width'], shape['height']
            mask[y:y+height, x:x+width] = 1
    return mask
```

```python
os.makedirs('masks', exist_ok=True)

for _id, annotation in annotations.items():
    image_path = os.path.join('/content/drive/MyDrive/Comic_Marvel', annotation['filename'])
    image = cv2.imread(image_path)
    if image is None:
        print(f"Image {annotation['filename']} not found. Skipping.")
        continue
    mask = create_mask(image.shape, annotation['regions'])
    mask_filename = os.path.splitext(annotation['filename'])[0] + '_mask.png'
    cv2.imwrite(os.path.join('masks', mask_filename), mask * 255)
```

```python
def save_chars(image_path, regions, output_dir, image_id):
    image = cv2.imread(image_path)
    if image is None:
        print(f"Image {image_path} not found. Skipping.")
        return

    for i, region in enumerate(regions):
        shape = region['shape_attributes']
        label = region['region_attributes']['type']
        x, y, width, height = shape['x'], shape['y'], shape['width'], shape['height']
        cropped = image[y:y+height, x:x+width]
        os.makedirs(os.path.join(output_dir, label), exist_ok=True)
        cv2.imwrite(os.path.join(output_dir, label, f'{image_id}_{i}.png'), cropped)

os.makedirs('characters', exist_ok=True)

for _id, annotation in annotations.items():
    img_path = os.path.join('/content/drive/MyDrive/Comic_Marvel', annotation['filename'])
    save_chars(img_path, annotation['regions'], 'characters',_id)
```
